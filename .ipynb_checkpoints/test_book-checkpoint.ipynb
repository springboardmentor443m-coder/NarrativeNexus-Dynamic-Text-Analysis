```python
from bs4 import BeautifulSoup
import re
```

```python
import os
os.listdir()
```

```python
def clean_html_file(file_path):

    # Read file
    with open(file_path, "r", encoding="utf-8") as f:
        html_content = f.read()

    # Use BeautifulSoup to parse HTML and remove tags
    soup = BeautifulSoup(html_content, "html.parser")

    # Remove <script> and <html> content completely
    for script in soup(["script", "html"]):
        script.extract()

    # Get visible text only
    text = soup.get_text(separator=" ")

    # Clean extra spaces and symbols
    text = re.sub(r'\s+', ' ', text).strip()

    return text

```

```python
file_path = "nlp_preprocessing_1.txt"  # 🔹 Your .txt or .html file name
cleaned_text = clean_html_file(file_path)

print("Cleaned Text:\n")
print(cleaned_text[:9000])  # Print first 9000 chars

```
